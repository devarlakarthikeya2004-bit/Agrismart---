```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (11, 6)

output = Path("charts")
output.mkdir(exist_ok=True)
```

```python
import ipywidgets as widgets
from IPython.display import display

price_upload = widgets.FileUpload(
    accept=".csv",
    multiple=False,
    description="Upload Prices"
)

arrival_upload = widgets.FileUpload(
    accept=".csv",
    multiple=False,
    description="Upload Arrivals"
)

display(price_upload)
display(arrival_upload)
```

```python
import io
import pandas as pd

def read_uploaded_csv(widget):
    files = widget.value

    if not files:
        raise ValueError("Please upload a CSV file first.")

    if isinstance(files, dict):
        file = next(iter(files.values()))
    else:
        file = files[0]

    return pd.read_csv(io.BytesIO(bytes(file["content"])))

prices = read_uploaded_csv(price_upload)
arrivals = read_uploaded_csv(arrival_upload)

print("Price dataset:")
display(prices.head())

print("Arrival dataset:")
display(arrivals.head())
```

```python
print("PRICE DATASET")
print("Rows and columns:", prices.shape)
print("Column names:", prices.columns.tolist())
display(prices.info())

print("\nARRIVAL DATASET")
print("Rows and columns:", arrivals.shape)
print("Column names:", arrivals.columns.tolist())
display(arrivals.info())
```

```python
print("Missing values in price dataset:")
display(prices.isnull().sum())

print("Missing values in arrival dataset:")
display(arrivals.isnull().sum())

print("Duplicate price rows:", prices.duplicated().sum())
print("Duplicate arrival rows:", arrivals.duplicated().sum())
```

```python
def standardize(df):
    df = df.copy()
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )
    return df

prices = standardize(prices)
arrivals = standardize(arrivals)

print("Price columns:", prices.columns.tolist())
print("Arrival columns:", arrivals.columns.tolist())
```

```python
def find_column(df, candidates):
    for candidate in candidates:
        if candidate in df.columns:
            return candidate
    raise ValueError(f"Column not found. Available columns: {df.columns.tolist()}")

price_date = find_column(prices, ["date", "arrival_date", "price_date"])
price_mandi = find_column(prices, ["mandi", "market", "market_name"])
price_value = find_column(
    prices,
    ["modal_price", "modal_price_rs_per_100kg", "price_rs_per_100kg", "price"]
)

arrival_date = find_column(arrivals, ["date", "arrival_date"])
arrival_mandi = find_column(arrivals, ["mandi", "market", "market_name"])
arrival_value = find_column(
    arrivals,
    ["arrival_tonnes", "arrivals_tonnes", "arrival", "arrivals"]
)

prices = prices.rename(columns={
    price_date: "Date",
    price_mandi: "Mandi",
    price_value: "Price"
})

arrivals = arrivals.rename(columns={
    arrival_date: "Date",
    arrival_mandi: "Mandi",
    arrival_value: "Arrival"
})

display(prices.head())
display(arrivals.head())
```

```python
prices["Date"] = pd.to_datetime(prices["Date"], errors="coerce")
arrivals["Date"] = pd.to_datetime(arrivals["Date"], errors="coerce")

prices["Mandi"] = prices["Mandi"].astype("string").str.strip().str.title()
arrivals["Mandi"] = arrivals["Mandi"].astype("string").str.strip().str.title()

prices["Price"] = pd.to_numeric(prices["Price"], errors="coerce")
arrivals["Arrival"] = pd.to_numeric(arrivals["Arrival"], errors="coerce")

print("Missing values before cleaning:")
display(prices[["Date", "Mandi", "Price"]].isna().sum())
display(arrivals[["Date", "Mandi", "Arrival"]].isna().sum())

prices = prices.dropna(subset=["Date", "Mandi", "Price"])
arrivals = arrivals.dropna(subset=["Date", "Mandi", "Arrival"])

prices = prices[prices["Price"] > 0]
arrivals = arrivals[arrivals["Arrival"] >= 0]

prices = prices.drop_duplicates()
arrivals = arrivals.drop_duplicates()

prices = prices.groupby(
    ["Date", "Mandi"], as_index=False
)["Price"].median()

arrivals = arrivals.groupby(
    ["Date", "Mandi"], as_index=False
)["Arrival"].sum()

print("Cleaned price records:", len(prices))
print("Cleaned arrival records:", len(arrivals))
```

```python
df = pd.merge(
    prices,
    arrivals,
    on=["Date", "Mandi"],
    how="inner",
    validate="one_to_one"
)

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%b")

df = df[df["Year"].between(2010, 2012)].copy()

print("Merged dataset:")
display(df.head())

print("Total records:", len(df))
print("Number of mandis:", df["Mandi"].nunique())
print("Date range:", df["Date"].min(), "to", df["Date"].max())
```

```python
summary = df.groupby("Mandi").agg(
    Average_Price=("Price", "mean"),
    Median_Price=("Price", "median"),
    Total_Arrival=("Arrival", "sum"),
    Average_Daily_Arrival=("Arrival", "mean"),
    Price_Std=("Price", "std"),
    Reporting_Days=("Date", "nunique")
).reset_index()

summary["Price_CV"] = (
    summary["Price_Std"] / summary["Average_Price"]
)

summary = summary.sort_values("Average_Price")

display(summary.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

output = Path("charts")
output.mkdir(exist_ok=True)
```

```python
price_summary = summary.sort_values("Average_Price")

plt.figure(figsize=(12, 7))

sns.barplot(
    data=price_summary,
    x="Average_Price",
    y="Mandi",
    hue="Mandi",
    palette="viridis",
    legend=False
)

plt.title("Average Tomato Prices Across UP Mandis (2010–2012)", fontsize=15)
plt.xlabel("Average Price (₹ per 100 kg)")
plt.ylabel("Mandi")
plt.tight_layout()

plt.savefig("charts/01_average_tomato_prices.png", dpi=300)
plt.show()
```

```python
arrival_summary = summary.sort_values("Total_Arrival", ascending=False)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=arrival_summary,
    x="Total_Arrival",
    y="Mandi",
    hue="Mandi",
    palette="crest",
    legend=False
)

plt.title("Total Tomato Arrivals Across UP Mandis (2010–2012)", fontsize=15)
plt.xlabel("Total Tomato Arrivals (Tonnes)")
plt.ylabel("Mandi")
plt.tight_layout()

plt.savefig("charts/02_total_tomato_arrivals.png", dpi=300)
plt.show()
```

```python
monthly_prices = (
    df.groupby(pd.Grouper(key="Date", freq="MS"))
    .agg(Average_Price=("Price", "mean"))
    .reset_index()
)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_prices,
    x="Date",
    y="Average_Price",
    marker="o",
    linewidth=2
)

plt.title("Monthly Tomato Price Trends in Uttar Pradesh (2010–2012)", fontsize=15)
plt.xlabel("Date")
plt.ylabel("Average Price (₹ per 100 kg)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("charts/03_monthly_price_trends.png", dpi=300)
plt.show()
```

```python
seasonal_prices = df.pivot_table(
    index="Mandi",
    columns="Month",
    values="Price",
    aggfunc="mean"
)

seasonal_prices = seasonal_prices.reindex(columns=range(1, 13))

seasonal_prices.columns = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(15, 8))

sns.heatmap(
    seasonal_prices,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Seasonal Tomato Price Heatmap (2010–2012)", fontsize=15)
plt.xlabel("Month")
plt.ylabel("Mandi")
plt.tight_layout()

plt.savefig("charts/04_seasonal_price_heatmap.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=summary,
    x="Average_Price",
    y="Average_Daily_Arrival",
    hue="Mandi",
    s=180
)

plt.title("Tomato Procurement: Price vs Arrival Analysis", fontsize=15)
plt.xlabel("Average Tomato Price (₹ per 100 kg)")
plt.ylabel("Average Daily Tomato Arrival (Tonnes)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()

plt.savefig("charts/05_price_vs_arrival.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(14, 7))

sns.boxplot(
    data=df,
    x="Mandi",
    y="Price",
    color="skyblue",
    showfliers=False
)

plt.title("Tomato Price Distribution Across UP Mandis", fontsize=15)
plt.xlabel("Mandi")
plt.ylabel("Tomato Price (₹ per 100 kg)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("charts/06_price_stability.png", dpi=300)
plt.show()
```

```python
ranking = summary.copy()

def score(series, higher_is_better=True):
    if series.max() == series.min():
        return pd.Series(0.5, index=series.index)
    normalized = (
        (series - series.min()) /
        (series.max() - series.min())
    )
    return normalized if higher_is_better else 1 - normalized

ranking["Price_Score"] = score(
    ranking["Average_Price"], False
)

ranking["Arrival_Score"] = score(
    ranking["Average_Daily_Arrival"], True
)

ranking["Stability_Score"] = score(
    ranking["Price_CV"].fillna(ranking["Price_CV"].median()), False
)

ranking["Final_Score"] = (
    0.40 * ranking["Price_Score"] +
    0.40 * ranking["Arrival_Score"] +
    0.20 * ranking["Stability_Score"]
)

ranking = ranking.sort_values("Final_Score", ascending=False)
ranking["Rank"] = range(1, len(ranking) + 1)

display(
    ranking[[
        "Rank",
        "Mandi",
        "Average_Price",
        "Average_Daily_Arrival",
        "Price_CV",
        "Final_Score"
    ]].round(3)
)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=ranking,
    x="Final_Score",
    y="Mandi",
    hue="Mandi",
    palette="viridis",
    legend=False
)

plt.title("Tomato Procurement Suitability Ranking", fontsize=15)
plt.xlabel("Weighted Procurement Score")
plt.ylabel("Mandi")
plt.tight_layout()

plt.savefig("charts/07_final_mandi_ranking.png", dpi=300)
plt.show()
```

```python
top_three = ranking.head(3)

print("TOP THREE TOMATO PROCUREMENT LOCATIONS")
print()

display(
    top_three[[
        "Rank",
        "Mandi",
        "Average_Price",
        "Average_Daily_Arrival",
        "Final_Score"
    ]].round(2)
)

print("Recommended procurement mandi:", ranking.iloc[0]["Mandi"])
```
